```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Almanya bayrağı

flag_germany=np.zeros((360,540,3),dtype=np.uint8)
```

```python
flag_germany.shape
```

```python
#siyah-kırmızı-sarı
flag_germany[:120,:,:]=[0,0,0]     #satırlardan 0-120 siyah, sütun sabit 
flag_germany[120:240,:,:]=[255,0,0]  #120-240 arası kırmızı
flag_germany[240:,:,:]=[240,220,0]    #240 sonrası sarı
```

```python
plt.imshow(flag_germany)
plt.axis("off")
```
