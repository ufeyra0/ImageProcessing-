```python
height,width,channel=150,210,3

my_black_img=np.zeros((height,width,channel),dtype=np.uint8)
my_white_img=np.ones((height,width,channel),dtype=np.uint8)*255     #beyaz olması için ones ile hepsini 1 yaptık sonra 255 ile çarp

#iki farklı yoldan gri rengini oluşturmak
my_gray_img_1=np.random.randint(127,128,size=(height,width,channel))
my_gray_img_2=np.ones((height,width,channel),dtype=np.uint8)*127       #gri olması için ones ile hepsini 1 yaptık sonra 127 ile çarp

#arka planı mor rengine boyuyoruz
fig=plt.figure(figsize=(10,7),facecolor="purple")

plt.subplot(2,2,1)
plt.imshow(my_black_img)
plt.axis("off")
plt.title("Black İmage")

plt.subplot(2,2,2)
plt.imshow(my_white_img)
plt.axis("off")
plt.title("White İmage ")

plt.subplot(2,2,3)
plt.imshow(my_gray_img_1)
plt.axis("off")
plt.title("Gray İmage 1")

plt.subplot(2,2,4)
plt.imshow(my_gray_img_2)
plt.axis("off")
plt.title("Gray İmage 2")


```

```python
#Red Green Blue renklerini oluşturup tek bir plot'ta gösterin

my_red_img=np.zeros((height,width,channel),dtype=np.uint8)
my_green_img=np.zeros((height,width,channel),dtype=np.uint8)
my_blue_img=np.zeros((height,width,channel),dtype=np.uint8)
```

```python
my_red_img[:,:,0]=255         #red image için ilk indeks[0] kanal rengini 255 yap
my_green_img[:,:,1]=255
my_blue_img[:,:,2]=255
```

```python
fig=plt.figure(figsize=(10,7))

plt.subplot(1,3,1)
plt.imshow(my_red_img)
plt.axis("off")
plt.title("Red İmage")

plt.subplot(1,3,2)
plt.imshow(my_green_img)
plt.axis("off")
plt.title("Green İmage ")

plt.subplot(1,3,3)
plt.imshow(my_blue_img)
plt.axis("off")
plt.title("Blue İmage ")
```

```python
my_new_black_img=my_red_img*0
my_new_white_img=my_red_img+my_green_img+my_blue_img

fig=plt.figure(figsize=(10,7),facecolor="purple")

plt.subplot(1,2,1)
plt.imshow(my_new_black_img)
plt.axis("off")
plt.title("Black İmage")

plt.subplot(1,2,2)
plt.imshow(my_new_white_img)
plt.axis("off")
plt.title("White İmage ")
```

```python
#tanımladığımız my_re,my_green;my_blue değişkenlerini kullanarak mor ve sarı renklerini elde ediniz

my_purple_img=my_red_img+my_blue_img
my_yellow_img=my_green_img+my_red_img

fig=plt.figure(figsize=(10,7),facecolor="orange")

plt.subplot(1,2,1)
plt.imshow(my_purple_img)
plt.axis("off")
plt.title("Purple İmage")

plt.subplot(1,2,2)
plt.imshow(my_yellow_img)
plt.axis("off")
plt.title("Yellow İmage ")
```

```python
#fenerbahçe renklerini çiziniz

my_blue_yellow_img=np.zeros((height,width,channel),dtype=np.uint8)
my_blue_yellow_img[:,:105,:]=[255,255,0]
my_blue_yellow_img[:,105:,:]=[0,0,255]

plt.imshow(my_blue_yellow_img)
plt.axis("off")
plt.title("Fenerbahçe Renkleri")

#Eğer bir görseli eşit şekilde parçalayıp farklı renklere boyamak istiyorsanız satır veya sütun bazında indeksleme yapmalısınız.
```

```python
#COLOR DETECTİON(renk tespit etme)= bu işlem için görüntünün tüm satır ve sütununda ve/veya istediğiniz 
#ya da tüm kanallarında gezmek durumundayız.

#Örneğin fenerbahçe renklerini gs renklerini dönüştürmek için önce fenerbahçe amblemindeki mavi rengini tespit etmemiz lazım
#sonra bunu kordinat sistemini çıkarmamız lazım sonra bu kordinat sistemini tekrar bir bütün indels olarak kullanıp bunun
#değerlerini kırmızı rengi olarak atamamız lazım

kordinat_mavi=(my_blue_yellow_img[:,:,0]==0)&(my_blue_yellow_img[:,:,1]==0)&(my_blue_yellow_img[:,:,2]==255)

#mavi renginin kordinatlarını tekrar fenerbahçe amblemine yerleştiriyoruz 
#bu defa karşınıza artık kordinat değil, o kordinattaki değerler geliyor,
#bu aşamadan sonra da mavi rengi yerine kırmızı renginin kodlarını yazıyoruz.
#yani [255,0,0]

my_blue_yellow_img[kordinat_mavi]=[255,0,0]

plt.imshow(my_blue_yellow_img)
plt.axis("off")
plt.title("Galatasaray Renkleri")
```

```python
# sarı renk beyaza kırmızı renk siyaha dönüşecek

kordinat_sari=(my_blue_yellow_img[:,:,0]==255)&(my_blue_yellow_img[:,:,1]==255)&(my_blue_yellow_img[:,:,2]==0)
kordinat_kirmizi=(my_blue_yellow_img[:,:,0]==255)&(my_blue_yellow_img[:,:,1]==0)&(my_blue_yellow_img[:,:,2]==0)

my_blue_yellow_img[kordinat_sari]=[255,255,255]
my_blue_yellow_img[kordinat_kirmizi]=[0,0,0]

fig=plt.figure(figsize=(10,7),facecolor="gray")

plt.subplot(1,1,1)
plt.imshow(my_blue_yellow_img)
plt.axis("off")
plt.title("Kara Kartal Renkleri",size=25)
```
